## Sex Hormones

Sex hormones also affect how the body stores adiposity. Hormonal imbalance can cause varying effects on the human body. For example, female with polycystic ovarian syndrome (PCOS) have higher testosterone levels. They are more likely to be overweight compared to female with normal hormone balance. 

This analysis will evaluate how the sex hormones affect the BMI.

NHANES data includes the following sex hormones and their ranges:
* 17-hydroxyprogesterone (17-OHP): a steroid hormone produced by the adrenal glands and gonads, acting as a precursor to cortisol.
    * Highly dependent on the time of day as well as the hormone cycle of women but generally <200-220 ng/dL for 
    both males and females
* Androstenedione ("andro"): a weak androgenic steroid hormone produced by the adrenal glands and gonads, acting as a direct precursor to testosterone and estrogen.
    * 27–152 ng/dL for males
    * 41–262 ng/dL for females
